## McCord 2006
### We didn't even know they were dunes yet. 

Composition of Titan's surface from Cassini VIMS

This paper is actually primarily about retrieving albedos. Or, well, reflectances that are surface I/F values and then using them to estimate the surface albedo. The key pieces of data are Table 6 and Figure 14.






The "ddark" terrain is the dunes terrain, though notably McCord didn't know that, it was just a dark terrain to this group. 

The important thing to note about this observation is that it is extreemly limited, only a single set of pixels from a very small area on Titan, taken at an angle and not directly above. Since McCord mentions no angle correction, it is very possible the iephi of 37.2, 25.5, 12.7 (phase angle is not azimuth, careful) simly wasn't corrected for. 

A good way to check for this correction is simply thus: the I/F would be lower than the surface albedo at an oblique angle. So what is the retrieved I/F, as opposed to the final albedo they report? Well, the value they report is the "best estimate" which is just an averaging of the three methods they used to actually get at the information. We see in Fig 13a that the filled line is essentially the same as what is finally observed, just more cluttered. All the methods are in play here, which is suspect.

To be clear: the three esimates are: no atmospheric correction, simple cosine extinction, and a radiative trasnfer model. One might argue that this means the best estimate from the radiative transfer model should be taken... but it only gets estimates for five windows, treating the double window as one! So yeah, if we really want all the windows, some averaging was required. The estimates did weight the radiative transfer model as superior to the others, and at longer wavelengths there is minimal disagreement anyway. But this means the short wavelength values for McCord are highly suspect. 

As these values are reported as direct I/F, it is pretty clear that no angular correction was mode. 

So, since the no atmosphere correction model was used, and it makes things brighter, the real values should be lower. But since there's no angle correction, the real values should be higher! Which is probably why the McCord values don't seem absurd when compared to the others; there are two competing offset effects! Wild!

Now to examine the radiative transfer code itself. It is from Griffith 2003, using approximations from Stamnes 1988 (ooh, an oldie). Absorption coefficients from Husson 1991. We have methane distribution functions that are assumed. The Voyager profile in Lellouch 1989 is used for the thermal profile. CIA is included (called pressure-induced here) and CO's 0-1 vibrational band. CO fluorescence is ignored. Mie calculations are used for scattering. Assuming Khare's optical constants for tholins. 40-80 km scale hight. 

Notably a wide range of optical depths fit the observations so error bars are huge on the radtran model.

Notably this is also looking at a single location; perhaps it just is at the *right* location. This is specifically a dune field in Shangri-La.


For completeness, let's examine Griffiths 2003 "Evidence for the Exposure of Water Ice on Titan's Surface", which is where the radtran model comes from. Which claims exposed icy bedrock everywhere. (Which... is suspect, but this is an old paper). Griffiths model ultimately comes from Stanmes 1988 as well. 

Examining STamnes 1988, it's a general non-Titan-specific radiative transfer for plane parallel situations.
