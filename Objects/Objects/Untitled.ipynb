The equation we have been using for the volume error in the code:

$$ \frac{r_e^3}{r^3} $$

But the true relative volume error is given by

$$ \frac{(r+r_e)^3}{r^3} - 1 $$

Which, yes, contains the top TERM, but is not actually the EXACT ERROR.

Actually using the proper equation to calculate the errors here kills about 100 objects from our tables. If it was just a bunch of tiny objects this would be no issue, but this includes, Sag A* and Halley's Comet.

In by-term form:

$$ \frac{3r^2r_e}{r^3} + \frac{3rr_e^2}{r^3} + \frac{r_e^3}{r^3} $$

$$ \frac{3r_e}{r} + \frac{3r_e^2}{r^2} + \frac{r_e^3}{r^3} $$
